```python
#Most popular tracks

top_songs = df.sort_values(by='Stream', ascending=False)

top_songs[['Track', 'Artist', 'Stream']].head(10)
```

```python
#Average stream

df['Stream'].mean()
```

```python
#Average stream by the Artist

df.groupby('Artist')['Stream'].mean().sort_values(ascending=False).head(10)
```

```python
#Top Artits (bar chart)

top_artists_avg_streams = df.groupby('Artist')['Stream'].mean().sort_values(ascending=False).head(10)
top_artists_avg_streams.sort_values().plot(kind="barh", figsize=(10,5))

plt.title("Top 10 Artists by Average Streams")
plt.xlabel("Average Streams")
plt.ylabel("Artist")
plt.show()
```

```python
#Danceability vs Stream relations

import matplotlib.pyplot as plt

plt.scatter(df['Danceability'], df['Stream'])
plt.xlabel("Danceability")
plt.ylabel("Stream")
plt.show()
```

```python
#Danceability vs Stream (boxplot)

sns.boxplot(x='Danceability', y='Stream', data=df)
```

```python
#Stream distribution (boxplot)

plt.boxplot(df['Stream'])
plt.title("Stream Distribution")
plt.show()
```

```python
#Duration Anlaysis

df['Duration_min'].describe()
```

```python
#Duration by the Artist

artist_duration = df.groupby('Artist')['Duration_min'].mean()
artist_duration.head()
```

```python
#Which artists are famous for longer songs

top_artists = artist_duration.sort_values(ascending=False).head(10)

top_artists.plot(kind='bar')
plt.title("Top Artists by Average Song Duration")
plt.ylabel("Minutes")
plt.show()
```

```python
# longest duration

artist_duration.sort_values(ascending=False).head(10)
```

```python
#Median duration

artist_duration_median = df.groupby('Artist')['Duration_min'].median()
```

```python
#duration by the Album_type

avg_duration = df.groupby('Album_type')['Duration_min'].mean()

avg_duration
avg_duration.sort_values(ascending=False)
```

```python
avg_duration.plot(kind='bar')

plt.title("Average Duration by Album Type")
plt.xlabel("Album Type")
plt.ylabel("Duration (minutes)")
plt.show()
```
